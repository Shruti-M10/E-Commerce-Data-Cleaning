## 6. Duplicate Record Analysis

Duplicate analysis is performed to identify fully duplicated records and repeated order identifiers.

Fully duplicated rows may indicate accidental copies, while repeated OrderID values require further investigation to determine whether they represent duplicate orders or valid repeated entries.

Only confirmed duplicate records will be removed.

```python
# Check fully duplicated rows
exact_duplicate_count = df.duplicated().sum()

print("Exact duplicate rows:", exact_duplicate_count)

# Check repeated OrderID values
repeated_order_id_rows = df[
    df.duplicated(subset="OrderID", keep=False)
].sort_values("OrderID")

print("Rows belonging to repeated OrderIDs:",
      len(repeated_order_id_rows))

print("Number of distinct repeated OrderIDs:",
      repeated_order_id_rows["OrderID"].nunique())

display(repeated_order_id_rows.head(10))
```

### Duplicate Analysis — Initial Findings

The duplicate analysis identified 20 fully duplicated rows. A separate check of the `OrderID` column found 39 rows associated with 19 distinct repeated order IDs.

These results indicate that exact duplicate rows and repeated order identifiers must be investigated separately. Repeated order IDs will not automatically be treated as errors because they may represent valid transactions or repeated entries.

**Cleaning decision:** Further inspection is required before deciding how to handle the repeated order IDs. No records have been removed at this stage.

```python
# Display all rows associated with repeated OrderIDs
display(
    df[df.duplicated(subset="OrderID", keep=False)]
    .sort_values("OrderID")
)
```

```python
order_id_counts = df["OrderID"].value_counts()

repeated_ids = order_id_counts[order_id_counts > 1]

display(
    repeated_ids.rename("Occurrence Count")
    .reset_index()
    .rename(columns={"index": "OrderID"})
)
```

```python
# Display all rows for repeated OrderIDs
repeated_orders = df[
    df.duplicated(subset="OrderID", keep=False)
].sort_values("OrderID")

display(repeated_orders)
```

```python
# Compare important fields for repeated OrderIDs
columns_to_check = [
    "OrderID",
    "CustomerID",
    "OrderDate",
    "ProductCategory",
    "ProductSubCategory",
    "UnitPrice",
    "Quantity",
    "TotalAmount",
    "FinalAmount"
]

display(repeated_orders[columns_to_check])
```

```python
# Check how many unique full records exist for each repeated OrderID
duplicate_check = (
    repeated_orders
    .drop_duplicates()
    .groupby("OrderID")
    .size()
)

print("Number of unique full records for each repeated OrderID:")
display(duplicate_check)

print(
    "Repeated OrderIDs with identical records:",
    (duplicate_check == 1).sum()
)

print(
    "Repeated OrderIDs with different records:",
    (duplicate_check > 1).sum()
)
```

### Duplicate Analysis — Final Findings

The duplicate analysis identified 20 fully duplicated rows in the original dataset. Additionally, 39 rows were associated with 19 distinct repeated `OrderID` values.

After removing exact duplicates for investigation, each of the 19 repeated order IDs had exactly one unique full record. No repeated order ID had different record versions.

**Cleaning decision:** The repeated entries are treated as duplicate copies because their full records are identical. Exact duplicate rows will be removed from the cleaned dataset while retaining one copy of each record. The raw dataset will remain unchanged.

## 7. Removing Exact Duplicate Rows

The investigation identified 20 exact duplicate rows in the original dataset. The repeated `OrderID` groups were also examined, and all 19 groups contained identical full records.

To prevent duplicate records from affecting future analysis, a separate copy of the original dataset will be created. Extra copies of exact duplicate rows will be removed, while one copy of each record will be retained. The original dataset will remain unchanged.

```python
# Create a separate copy of the original dataset
df_clean = df.copy()

# Remove exact duplicate rows from the copy
df_clean = df_clean.drop_duplicates().copy()

# Check the results
print("Original dataset rows:", len(df))
print("Cleaned dataset rows:", len(df_clean))
print("Duplicate rows remaining:", df_clean.duplicated().sum())
```

### Duplicate Removal — Results

The original dataset contained 2,000 rows, including 20 extra copies of exact duplicate records. After applying `drop_duplicates()` to a separate copy of the dataset, the cleaned DataFrame contains 1,980 rows.

A verification check confirmed that zero exact duplicate rows remain in `df_clean`.

The original DataFrame, `df`, was preserved without modification. The cleaned DataFrame, `df_clean`, will be used for the remaining validation and cleaning steps.

## 8. Data Validation

Data validation checks whether values follow reasonable rules and expected formats. In this step, we will examine dates, quantities, prices, discounts, transaction amounts, shipping days, review scores, and categorical values.

The checks will help us identify potential data-quality issues before deciding whether any additional cleaning is necessary.

```python
# Check data types and basic statistics
print("Data types and non-null counts:")
df_clean.info()

print("\nNumeric column summary:")
display(df_clean.describe())

print("\nUnique values in selected categorical columns:")
for column in ["Returned", "OrderSeason", "OrderDayOfWeek",
                "PaymentMethod", "DeviceType", "MembershipTier"]:
    print(f"\n{column}:")
    print(df_clean[column].value_counts(dropna=False))
```

### 8.1 Order Date Validation

The `OrderDate` column is currently stored as a string (`str`) rather than a date type. Date validation will check whether the values can be interpreted as valid dates.

The conversion will first be tested on a separate Series so that invalid date values can be identified before modifying the cleaned dataset.

```python
# Test whether OrderDate values can be parsed as dates
parsed_dates = pd.to_datetime(
    df_clean["OrderDate"],
    format="%m/%d/%Y",
    errors="coerce"
)

# Count dates that could not be parsed
invalid_date_count = parsed_dates.isna().sum()

print("Invalid dates:", invalid_date_count)

if invalid_date_count == 0:
    print("Earliest order date:", parsed_dates.min())
    print("Latest order date:", parsed_dates.max())
else:
    print("Rows with invalid dates:")
    display(df_clean.loc[parsed_dates.isna(), ["OrderID", "OrderDate"]])
```

### Order Date Validation — Findings

The `OrderDate` column was tested using the date format `%m/%d/%Y`.

- Invalid dates found: 0
- Earliest order date: January 1, 2022
- Latest order date: February 28, 2025

**Cleaning decision:** All order dates were successfully parsed. No records need to be removed because of invalid dates. The column will be converted to a proper datetime data type for easier sorting and date-based analysis.

### 8.2 Transaction Amount Validation

The transaction amount columns will be checked to determine whether the recorded totals and discounts are consistent with the unit price, quantity, and discount percentage.

For each record, the expected total amount is calculated as `UnitPrice × Quantity`. The expected discount is calculated from the total amount and discount percentage, and the expected final amount is calculated by subtracting the discount amount from the total amount.

Any differences will be investigated before deciding whether corrections are necessary.

```python
# Calculate expected transaction amounts
expected_total = df_clean["UnitPrice"] * df_clean["Quantity"]

expected_discount = (
    expected_total * df_clean["DiscountPercent"] / 100
).round(2)

expected_final = expected_total - expected_discount

# Compare calculated amounts with recorded amounts
total_mismatch = ~(
    (df_clean["TotalAmount"] - expected_total).abs() < 0.011
)

discount_mismatch = ~(
    (df_clean["DiscountAmount"] - expected_discount).abs() < 0.011
)

final_mismatch = ~(
    (df_clean["FinalAmount"] - expected_final).abs() < 0.011
)

print("TotalAmount mismatches:", total_mismatch.sum())
print("DiscountAmount mismatches:", discount_mismatch.sum())
print("FinalAmount mismatches:", final_mismatch.sum())
```

### Transaction Amount Validation — Findings

The transaction amount validation was performed on all 1,980 rows in the cleaned DataFrame.

- `TotalAmount` mismatches: 0
- `DiscountAmount` mismatches: 0
- `FinalAmount` mismatches: 0

The recorded total amounts matched `UnitPrice × Quantity`. The recorded discount amounts matched the calculated discounts, and the final amounts matched the total amounts minus the discounts, allowing for rounding.

**Cleaning decision:** No corrections were required for these three transaction amount columns because all records passed the tested calculations.

### 8.3 Numeric Value Validation

The numeric columns will be checked for values that violate basic data-quality rules. Quantities should be positive, unit prices should be greater than zero, discount percentages should be between 0 and 100, shipping days should not be negative, and review scores should be within the expected range of 1 to 5.

The results will determine whether any numeric values require further investigation.

```python
# Check for invalid numeric values
print("Quantity <= 0:", (df_clean["Quantity"] <= 0).sum())
print("UnitPrice <= 0:", (df_clean["UnitPrice"] <= 0).sum())
print(
    "DiscountPercent outside 0–100:",
    (
        (df_clean["DiscountPercent"] < 0) |
        (df_clean["DiscountPercent"] > 100)
    ).sum()
)
print("ShippingDays < 0:", (df_clean["ShippingDays"] < 0).sum())
print(
    "ReviewScore outside 1–5:",
    (
        (df_clean["ReviewScore"] < 1) |
        (df_clean["ReviewScore"] > 5)
    ).sum()
)
```

### Numeric Value Validation — Findings

The numeric validation checks were performed on all 1,980 records in the cleaned DataFrame.

- Invalid quantities (less than or equal to zero): 0
- Invalid unit prices (less than or equal to zero): 0
- Discount percentages outside the 0–100 range: 0
- Negative shipping durations: 0
- Review scores outside the 1–5 range: 0

**Cleaning decision:** No corrections were required for these numeric columns because all records passed the specified validation checks.

### 8.4 Categorical Value Consistency

Categorical columns will be inspected for inconsistent spelling, capitalization, whitespace, and unexpected labels. This includes return status, seasons, days of the week, product categories, payment methods, device types, and membership tiers.

Values will not be changed unless the inspection identifies a clear inconsistency.

```python
# Inspect distinct values in categorical columns
categorical_columns = [
    "Returned",
    "OrderSeason",
    "OrderDayOfWeek",
    "ProductCategory",
    "ProductSubCategory",
    "PaymentMethod",
    "DeviceType",
    "MembershipTier"
]

for column in categorical_columns:
    print(f"\n--- {column} ---")
    print(sorted(df_clean[column].dropna().unique()))
```

### Categorical Value Consistency — Initial Findings

The distinct values in the categorical columns were inspected for inconsistent spelling, capitalization, and unexpected labels.

- `Returned` contains `Yes` and `No`.
- `OrderSeason` contains Fall, Spring, Summer, and Winter.
- `OrderDayOfWeek` contains all seven days of the week.
- `ProductCategory`, `ProductSubCategory`, `PaymentMethod`, `DeviceType`, and `MembershipTier` contain clearly formatted category labels.

**Initial cleaning decision:** No obvious spelling or capitalization inconsistencies were found in the displayed distinct values. Further checks will be performed for extra whitespace and consistency between the order dates, seasons, and weekdays.

```python
# Check text columns for leading or trailing spaces
text_columns = df_clean.select_dtypes(include=["str"]).columns

for column in text_columns:
    has_extra_spaces = df_clean[column].apply(
        lambda value: isinstance(value, str) and value != value.strip()
    )

    if has_extra_spaces.any():
        print(f"{column}: {has_extra_spaces.sum()} values with extra spaces")

print("Whitespace check completed.")
```

### Whitespace Validation — Findings

All text columns were checked for leading and trailing whitespace.

- Columns with extra whitespace: None detected.

**Cleaning decision:** No whitespace removal was required because the check did not identify any values with leading or trailing spaces. The dataset remains unchanged by this check.

```python
# Convert OrderDate to datetime for validation
parsed_dates = pd.to_datetime(
    df_clean["OrderDate"],
    format="%m/%d/%Y",
    errors="coerce"
)

# Calculate the expected weekday and season
expected_weekday = parsed_dates.dt.day_name()

month = parsed_dates.dt.month

expected_season = month.map({
    12: "Winter", 1: "Winter", 2: "Winter",
    3: "Spring", 4: "Spring", 5: "Spring",
    6: "Summer", 7: "Summer", 8: "Summer",
    9: "Fall", 10: "Fall", 11: "Fall"
})

# Compare recorded and expected values
weekday_mismatch = df_clean["OrderDayOfWeek"] != expected_weekday
season_mismatch = df_clean["OrderSeason"] != expected_season

print("Weekday mismatches:", weekday_mismatch.sum())
print("Season mismatches:", season_mismatch.sum())

print("\nWeekday mismatch examples:")
display(df_clean.loc[
    weekday_mismatch,
    ["OrderDate", "OrderDayOfWeek"]
].head(10))

print("\nSeason mismatch examples:")
display(df_clean.loc[
    season_mismatch,
    ["OrderDate", "OrderSeason"]
].head(10))
```

### Date and Category Consistency — Findings

The recorded weekdays and seasons were compared with the values calculated from `OrderDate`.

- Weekday mismatches: 0
- Season mismatches: 0

All recorded weekdays match the actual weekdays calculated from the order dates. All recorded seasons match the defined meteorological seasons.

**Cleaning decision:** No corrections were required for `OrderDayOfWeek` or `OrderSeason` because all records passed both validation checks.

```python
# Inspect distinct values in the remaining text columns
remaining_text_columns = [
    "FirstName",
    "LastName",
    "State",
    "City"
]

for column in remaining_text_columns:
    print(f"\n--- {column} ---")
    print("Number of unique values:", df_clean[column].nunique())
    print("Sample values:")
    print(sorted(df_clean[column].dropna().unique())[:30])
```

```python
# Check name and location columns for formatting inconsistencies
columns_to_check = ["FirstName", "LastName", "State", "City"]

for column in columns_to_check:
    values = df_clean[column].dropna()

    # Check for leading/trailing spaces
    spaces = values[values != values.str.strip()]

    # Check for values that are not in title case,
    # except State, which should use uppercase codes
    if column == "State":
        inconsistent_case = values[values != values.str.upper()]
    else:
        inconsistent_case = values[values != values.str.title()]

    print(f"\n--- {column} ---")
    print("Values with extra spaces:", len(spaces))
    print("Values with inconsistent capitalization:", len(inconsistent_case))

    if len(spaces) > 0:
        print("Examples with extra spaces:", spaces.unique()[:10])

    if len(inconsistent_case) > 0:
        print("Examples with inconsistent capitalization:",
              inconsistent_case.unique()[:10])
```

### Name and Location Formatting — Findings

The `FirstName`, `LastName`, `State`, and `City` columns were checked for leading/trailing spaces and inconsistent capitalization.

- Extra spaces detected: 0 across all four columns.
- Inconsistent capitalization detected: 0 across all four columns.

State codes use uppercase formatting, while names and city values follow title case.

**Cleaning decision:** No formatting corrections were required for these columns. The original values were retained.
